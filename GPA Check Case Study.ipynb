#GPA Check Case Study
#This app will accept a student's last name, first name, and GPA to determine if they have made the Dean's List, Honor Roll, or neither.

```python
while True:
    lastName = input("Enter last name [or ZZZ to exit]: ")
    if lastName == "ZZZ":
        print("Program exited.")
        break
    
    firstName = input("Enter first name: ")
    gpa = float(input("Enter GPA: "))
    name = firstName.capitalize() + " "+ lastName.capitalize()
    
    if gpa >= 3.5:
        print(name, "has made the Dean's List.")
    elif gpa >= 3.25:
        print(name, "has made the Honor Roll.")
    else:
        print(name, "has not made the Dean's List or the Honor Roll.") 
```
