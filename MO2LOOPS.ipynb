```python
secret = 8
guess = 6

if guess > secret:
    print("Too high")
elif guess < secret:
    print("Too low")
else:
    print("Just right")
```

```python
small = False
green = False

if small:
    if green:
        print("Pea")
    else:
        print("Cherry")
else:
    if green:
        print("Watermelon")
    else:
        print("Pumpkin")
```

```python
guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("Too low!")
    elif number == guess_me:
        print("Found it!")
        break
    else:
        print("Oops!")
        break
    number += 1
```

```python
guess_me = 5

for number in range(10):
    if number < guess_me:
        print("Too low!")
    elif number == guess_me:
        print("Found it!")
        break
    else:
        print("Oops!")
        break
```
